## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd
import requests
import io

# Load the starter data directly from the public repo
url = 'https://raw.githubusercontent.com/flyrank-bih/flyrank-ml-internship-starter/main/data/processed/refresh_feature_vector.csv'

try:
    response = requests.get(url)
    response.raise_for_status() # Raise an HTTPError for bad responses (4xx or 5xx)
    csv_content = response.text
    df = pd.read_csv(io.StringIO(csv_content))

    print(f"Total pages analyzed in dataset: {len(df)}")
    print(f"Pages actually declining: {df['is_declining'].sum()}")
    print(f"Decline rate: {(df['is_declining'].mean() * 100):.1f}%")

except requests.exceptions.RequestException as e:
    print(f"Error fetching data: {e}")
    print("Please check the URL or your internet connection.")
except Exception as e:
    print(f"An unexpected error occurred: {e}")
```
